```python
def clean_time(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Normal HH:MM time
    if ":" in value:
        return pd.to_datetime(value, format="%H:%M", errors="coerce").time()

    # Excel fractional-day time
    try:
        fraction = float(value)

        if 0 <= fraction < 1:
            total_minutes = round(fraction * 24 * 60)
            hours = (total_minutes // 60) % 24
            minutes = total_minutes % 60

            return pd.Timestamp(
                year=2000,
                month=1,
                day=1,
                hour=hours,
                minute=minutes
            ).time()

    except ValueError:
        pass

    return pd.NaT
```

```python
print(clean_time("21:55"))
print(clean_time("0.5"))
print(clean_time("0.75"))
print(clean_time("0.875"))
```

```python
from datetime import time

def clean_time(value):
    if pd.isna(value):
        return None

    value = str(value).strip()

    if ":" in value:
        try:
            return pd.to_datetime(
                value,
                format="%H:%M",
                errors="raise"
            ).time()
        except:
            return None

    try:
        fraction = float(value)

        if 0 <= fraction < 1:
            total_minutes = round(fraction * 1440)
            hours = (total_minutes // 60) % 24
            minutes = total_minutes % 60

            return time(hours, minutes)

    except:
        return None

    return None
```

```python
print(clean_time("21:55"))
print(clean_time("0.5"))
print(clean_time("0.75"))
print(clean_time("0.875"))
```

```python
def make_datetime(date_value, time_value):
    if pd.isna(time_value):
        return pd.NaT

    base_date = pd.Timestamp(date_value).normalize()
    value = str(time_value).strip()

    try:
        # Case 1: Excel fractional-day value
        if ":" not in value:
            fraction = float(value)

            if 0 <= fraction <= 1:
                return base_date + pd.to_timedelta(
                    fraction * 24,
                    unit="h"
                )

        # Case 2: HH:MM or HH:MM:SS
        parts = value.split(":")

        hour = int(parts[0])
        minute = int(parts[1])
        second = int(float(parts[2])) if len(parts) > 2 else 0

        # timedelta automatically handles 24:05 as next day 00:05
        return (
            base_date
            + pd.Timedelta(hours=hour)
            + pd.Timedelta(minutes=minute)
            + pd.Timedelta(seconds=second)
        )

    except:
        return pd.NaT
```

```python
df["Order_DateTime"] = df.apply(
    lambda row: make_datetime(
        row["Order_Date"],
        row["Time_Orderd"]
    ),
    axis=1
)

df["Pickup_DateTime"] = df.apply(
    lambda row: make_datetime(
        row["Order_Date"],
        row["Time_Order_picked"]
    ),
    axis=1
)
```
